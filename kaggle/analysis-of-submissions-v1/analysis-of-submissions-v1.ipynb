# general analysis of several Submissions , <70

```python
import pandas as pd

# 
file_paths = [
    '/kaggle/input/ergo-analysis/submission1.csv',
    '/kaggle/input/ergo-analysis/submission2.csv',
    '/kaggle/input/ergo-analysis/submission3.csv',
    '/kaggle/input/ergo-analysis/submission4.csv',
    '/kaggle/input/ergo-analysis/submission5.csv'
]


dataframes = [pd.read_csv(file_path) for file_path in file_paths]


for i, df in enumerate(dataframes, start=1):
    print(f" submission{i}.csv:")
    print(df.head(), '\n')  

```

```python

scores = [67.96, 68.05, 68.20, 68.63, 67.62]
```

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(len(dataframes), 2, figsize=(12, 5 * len(dataframes)))

for i, df in enumerate(dataframes):
  
    axes[i, 0].hist(df['target'], bins=50, color='blue', edgecolor='black')
    axes[i, 0].set_title(f' hist submission{i+1}.csv')
    axes[i, 0].set_xlabel('Value')
    axes[i, 0].set_ylabel('Frequency')
    axes[i, 0].grid(True)

    # Box plot for submission{i+1}.csv file
    axes[i, 1].boxplot(df['target'], vert=False, patch_artist=True)
    axes[i, 1].set_title(f'Box plot for submission{i+1}.csv')
    axes[i, 1].set_xlabel('Value')

plt.tight_layout()
plt.show()

```

```python
colors = ['blue', 'green', 'red', 'purple', 'orange']
markers = ['o', 's', '^', 'D', '*']

# Creating a scatter plot
plt.figure(figsize=(15, 10))
for i, df in enumerate(dataframes):
    label = f'submission{i+1} score {scores[i]}'
    plt.scatter(df['row_id'], df['target'], alpha=0.5, color=colors[i], marker=markers[i], label=label)

plt.title('Scatter Plot of "target" Column Values for All Files')
plt.xlabel('Row ID')
plt.ylabel('Value')
plt.legend()
plt.show()
```

```python
percent_outliers = []
count_outliers = []

for df in dataframes:
    Q1 = df['target'].quantile(0.25)
    Q3 = df['target'].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = df[(df['target'] < lower_bound) | (df['target'] > upper_bound)]
    percent_outliers.append((len(outliers) / len(df)) * 100)
    count_outliers.append(len(outliers))

# Calculating and collecting data on outliers
fig, axes = plt.subplots(1, 2, figsize=(15, 6))

# Plotting the percentage of outliers in each file
axes[0].bar(range(1, 6), percent_outliers, color='blue')
axes[0].set_title('Percentage of Outliers in Each File')
axes[0].set_xlabel('File Number')
axes[0].set_ylabel('Percentage of Outliers')
axes[0].set_xticks(range(1, 6))

# Plotting the count of outliers in each file
axes[1].bar(range(1, 6), count_outliers, color='green')
axes[1].set_title('Count of Outliers in Each File')
axes[1].set_xlabel('File Number')
axes[1].set_ylabel('Count of Outliers')
axes[1].set_xticks(range(1, 6))

plt.show()
```

```python
percent_sum_outliers = []

for df in dataframes:
    Q1 = df['target'].quantile(0.25)
    Q3 = df['target'].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = df[(df['target'] < lower_bound) | (df['target'] > upper_bound)]
    percent_sum_outliers.append((outliers['target'].sum() / df['target'].sum()) * 100)

# Creating a subplot for the remaining graph
fig, axes = plt.subplots(1, 1, figsize=(10, 6))

# Plotting the percentage of sum outliers relative to the total sum in each file
axes.bar(range(1, 6), percent_sum_outliers, color='green')
axes.set_title('Percentage of Sum Outliers Relative to Total Sum in Each File')
axes.set_xlabel('File Number')
axes.set_ylabel('Percentage of Total Sum')
axes.set_xticks(range(1, 6))

plt.show()
```

```python
sums = []
outlier_sums = []

for df in dataframes:
    Q1 = df['target'].quantile(0.25)
    Q3 = df['target'].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = df[(df['target'] < lower_bound) | (df['target'] > upper_bound)]
    outlier_sums.append(outliers['target'].sum())
    sums.append(df['target'].sum())

# Calculating percentages for outliers and normal data
percent_outliers = [outlier / total * 100 for outlier, total in zip(outlier_sums, sums)]
percent_normal = [100 - percent for percent in percent_outliers]

# Creating a subplot
fig, ax = plt.subplots(figsize=(10, 6))
bar_width = 0.4
index = range(1, 6)

# Plotting the percentage of normal data
ax.bar(index, percent_normal, bar_width, label='Normal Data')

# Plotting the percentage of outliers (in red) stacked on top of normal data
ax.bar(index, percent_outliers, bar_width, bottom=percent_normal, label='Outliers', color='red')

ax.set_xlabel('File Number')
ax.set_ylabel('Percentages')
ax.set_title('Ratio of Outliers to Normal Data in Each File')
ax.set_xticks(index)
ax.set_ylim(0, 100)
ax.legend()

plt.show()
```
